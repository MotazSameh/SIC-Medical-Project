```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import time

data = []

all_categories = [
    ("امراض الجهاز التنفسي", "امراض-الجهاز-التنفسي")
]

for category_name, category_slug in all_categories:

    print(f"Scraping {category_name}")

    page = 1
    collected = 0

    while collected < 1000:

        url = f"https://altibbi.com/اسئلة-طبية/{category_slug}?page={page}"

        response = requests.get(url)
        soup = BeautifulSoup(response.text, "html.parser")

        questions = soup.find_all(class_="question-item-container")

        if not questions:
            break

        for q in questions:

            if collected >= 1000:
                break

            question = q.find(class_="question-text mt-17 id-5-5")
            answer = q.find(class_="hide")

            data.append([
                category_name,
                question.get_text(separator=" ") if question else "",
                answer.get_text(separator=" ") if answer else ""
            ])

            collected += 1

        page += 1
        time.sleep(1)

    print(f"Collected {collected} questions")
```

```python
data_df = pd.DataFrame(data, columns=['category', 'question', 'answer'])
data_df.to_csv('data.csv', index=False)
print(len(data))
data_df.head()
```

```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup

session = requests.Session()

start_url = ("https://altibbi.com/%D8%A7%D8%B3%D8%A6%D9%84%D8%A9-%D8%B7%D8%A8%D9%8A%D8%A9/"
             "%D8%A7%D9%85%D8%B1%D8%A7%D8%B6-%D8%A7%D9%84%D8%AC%D9%87%D8%A7%D8%B2-"
             "%D8%A7%D9%84%D8%AA%D9%86%D9%81%D8%B3%D9%8A")

response = session.get(start_url, timeout=30)
soup_categories = BeautifulSoup(response.text, "html.parser")
```

```python
select_element_all = soup_categories.find("select", {"id": "w1"})

all_categories = []
if select_element_all:
    options_all = select_element_all.find_all("option")
    all_categories = [
        (option.get_text(strip=True), option["value"])
        for option in options_all if option.get("value")
    ]
print("Number of categories:", len(all_categories))
```

```python
def get_full_question(q_tag):
    """لو السؤال مقصوص (بينتهي بـ ...) بنجيبه كامل من صفحة السؤال."""
    text = q_tag.get_text(separator=" ")

    if not text.strip().endswith("..."):
        return text                  

    try:
        r = session.get(q_tag["href"], timeout=30)
        body = BeautifulSoup(r.text, "html.parser").select_one("div.qv-question-body")
        if body:
            return body.get_text(separator=" ")
    except requests.RequestException:
        pass

    return text 
```

```python
data = []

for category_name, category_slug in all_categories:

    print(f"Scraping {category_name}")

    page = 1
    collected = 0

    while collected < 1000:

        url = f"https://altibbi.com/اسئلة-طبية/{category_slug}?page={page}"

        try:
            response = session.get(url, timeout=30)
        except requests.RequestException as e:
            print(f"  Request failed on page {page}: {e}")
            break

        soup = BeautifulSoup(response.text, "html.parser")
        questions = soup.find_all(class_="question-item-container")

        if not questions:
            break

        for q in questions:

            if collected >= 1000:
                break

            question = q.find(class_="question-text mt-17 id-5-5")
            answer = q.find(class_="hide")

            data.append([
                category_name,
                get_full_question(question) if question else "",
                answer.get_text(separator=" ") if answer else ""
                
            ])

            collected += 1

        page += 1
        
        data_df = pd.DataFrame(data, columns=["category", "question", "answer"])
        data_df.to_csv("data.csv", index=False, encoding="utf-8-sig")

        time.sleep(1)

    print(f"Collected {collected} questions")


data_df = pd.DataFrame(data, columns=["category", "question", "answer"])
data_df.to_csv("data.csv", index=False, encoding="utf-8-sig")

print(len(data_df))
data_df.head()
```
